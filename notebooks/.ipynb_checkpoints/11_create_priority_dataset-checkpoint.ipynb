```python
import random
from pathlib import Path

import numpy as np
import pandas as pd
```

```python
BASE_DIR = Path.cwd().parent

PROCESSED_DATA_DIR = BASE_DIR / "data" / "processed"
PROCESSED_DATA_DIR.mkdir(parents=True, exist_ok=True)

NUM_SAMPLES = 50000

disaster_types = [
    "Flood",
    "Fire",
    "Earthquake",
    "Landslide",
    "Cyclone"
]

road_accessibility = [
    "Accessible",
    "Partially Accessible",
    "Not Accessible"
]

weather_conditions = [
    "Clear",
    "Rain",
    "Heavy Rain",
    "Thunderstorm",
    "Cyclone"
]

location_types = [
    "Urban",
    "Suburban",
    "Rural",
    "Forest",
    "Coastal",
    "Mountain",
    "Industrial"
]

incident_times = [
    "Morning",
    "Afternoon",
    "Evening",
    "Night"
]

damage_levels = [
    "Low",
    "Medium",
    "High"
]

yes_no = [
    "Yes",
    "No"
]
```

```python
print(NUM_SAMPLES)
```

```python
def calculate_priority_score(
    severity,
    people,
    injured,
    children,
    elderly,
    road,
    weather,
    time_since,
    disaster,
    location,
    communication,
    infrastructure,
    evacuation
):
    score = 0

    # Severity
    score += severity * 35

    # Number of people
    score += min((people / 250) * 20, 20)

    # Injured
    score += min((injured / 120) * 15, 15)

    # Children
    score += min(children * 0.25, 8)

    # Elderly
    score += min(elderly * 0.25, 8)

    # Road Accessibility
    road_score = {
        "Accessible": 0,
        "Partially Accessible": 3,
        "Not Accessible": 6
    }

    score += road_score[road]

    # Weather
    weather_score = {
        "Clear": 0,
        "Rain": 1,
        "Heavy Rain": 2,
        "Thunderstorm": 3,
        "Cyclone": 4
    }

    score += weather_score[weather]

    # Time Since Incident
    score += min((time_since / 180) * 4, 4)

    # Bonus Rules
    if severity > 0.90 and road == "Not Accessible":
        score += 5

    if people > 100 and injured > 40:
        score += 5

    if disaster == "Cyclone" and location == "Coastal":
        score += 4

    if disaster == "Earthquake" and location == "Urban":
        score += 3

    if disaster == "Flood" and weather == "Heavy Rain":
        score += 2

    if time_since > 60 and injured > 20:
        score += 4

    if communication == "No":
        score += 2

    if infrastructure == "High":
        score += 3

    if evacuation == "Yes":
        score += 2

    score = round(max(0, min(score, 100)), 2)

    if score < 30:
        priority = "Low"
    elif score < 50:
        priority = "Medium"
    elif score < 75:
        priority = "High"
    else:
        priority = "Critical"

    return score, priority
```

```python
def generate_flood():

    severity = round(random.uniform(0.60, 1.00), 2)

    people = random.randint(30, 150)

    children = random.randint(0, min(25, people))

    elderly = random.randint(0, min(25, people - children))

    injured = random.randint(
        max(2, int(people * 0.05)),
        max(5, int(people * 0.35))
    )

    road = random.choices(
        road_accessibility,
        weights=[10, 30, 60]
    )[0]

    weather = random.choices(
        ["Rain", "Heavy Rain", "Thunderstorm"],
        weights=[20, 50, 30]
    )[0]

    location = random.choice(
        ["Urban", "Rural", "Coastal"]
    )

    time_since = random.randint(5, 120)

    incident_time = random.choice(incident_times)

    infrastructure = random.choices(
        damage_levels,
        weights=[20, 40, 40]
    )[0]

    communication = random.choices(
        yes_no,
        weights=[30, 70]
    )[0]

    evacuation = random.choices(
        yes_no,
        weights=[80, 20]
    )[0]

    score, priority = calculate_priority_score(
        severity,
        people,
        injured,
        children,
        elderly,
        road,
        weather,
        time_since,
        "Flood",
        location,
        communication,
        infrastructure,
        evacuation
    )

    return {
        "Severity": severity,
        "Disaster_Type": "Flood",
        "People_Count": people,
        "Children_Count": children,
        "Elderly_Count": elderly,
        "Injured_Count": injured,
        "Road_Accessibility": road,
        "Weather_Condition": weather,
        "Location_Type": location,
        "Time_Since_Incident": time_since,
        "Incident_Time": incident_time,
        "Infrastructure_Damage": infrastructure,
        "Communication_Available": communication,
        "Evacuation_Required": evacuation,
        "Priority_Score": score,
        "Priority": priority
    }
```

```python
def generate_fire():

    severity = round(random.uniform(0.40, 0.90), 2)

    people = random.randint(5, 50)

    children = random.randint(0, min(10, people))

    elderly = random.randint(0, min(10, people - children))

    injured = random.randint(
        max(2, int(people * 0.20)),
        max(5, int(people * 0.70))
    )

    road = random.choices(
        road_accessibility,
        weights=[70, 20, 10]
    )[0]

    weather = random.choice(
        ["Clear", "Rain"]
    )

    location = random.choice(
        ["Urban", "Industrial", "Forest"]
    )

    time_since = random.randint(1, 40)

    incident_time = random.choice(incident_times)

    infrastructure = random.choices(
        damage_levels,
        weights=[40, 40, 20]
    )[0]

    communication = "Yes"

    evacuation = random.choices(
        yes_no,
        weights=[60, 40]
    )[0]

    score, priority = calculate_priority_score(
        severity,
        people,
        injured,
        children,
        elderly,
        road,
        weather,
        time_since,
        "Fire",
        location,
        communication,
        infrastructure,
        evacuation
    )

    return {
        "Severity": severity,
        "Disaster_Type": "Fire",
        "People_Count": people,
        "Children_Count": children,
        "Elderly_Count": elderly,
        "Injured_Count": injured,
        "Road_Accessibility": road,
        "Weather_Condition": weather,
        "Location_Type": location,
        "Time_Since_Incident": time_since,
        "Incident_Time": incident_time,
        "Infrastructure_Damage": infrastructure,
        "Communication_Available": communication,
        "Evacuation_Required": evacuation,
        "Priority_Score": score,
        "Priority": priority
    }
```

```python
def generate_earthquake():

    severity = round(random.uniform(0.75, 1.00), 2)

    people = random.randint(40, 200)

    children = random.randint(0, min(30, people))

    elderly = random.randint(0, min(30, people - children))

    injured = random.randint(
        max(5, int(people * 0.15)),
        max(15, int(people * 0.60))
    )

    road = random.choices(
        road_accessibility,
        weights=[0, 25, 75]
    )[0]

    weather = random.choice(
        ["Clear", "Rain"]
    )

    location = random.choice(
        ["Urban", "Mountain", "Rural"]
    )

    time_since = random.randint(5, 180)

    incident_time = random.choice(incident_times)

    infrastructure = random.choices(
        damage_levels,
        weights=[10, 30, 60]
    )[0]

    communication = random.choices(
        yes_no,
        weights=[40, 60]
    )[0]

    evacuation = "Yes"

    score, priority = calculate_priority_score(
        severity,
        people,
        injured,
        children,
        elderly,
        road,
        weather,
        time_since,
        "Earthquake",
        location,
        communication,
        infrastructure,
        evacuation
    )

    return {
        "Severity": severity,
        "Disaster_Type": "Earthquake",
        "People_Count": people,
        "Children_Count": children,
        "Elderly_Count": elderly,
        "Injured_Count": injured,
        "Road_Accessibility": road,
        "Weather_Condition": weather,
        "Location_Type": location,
        "Time_Since_Incident": time_since,
        "Incident_Time": incident_time,
        "Infrastructure_Damage": infrastructure,
        "Communication_Available": communication,
        "Evacuation_Required": evacuation,
        "Priority_Score": score,
        "Priority": priority
    }
```

```python
def generate_landslide():

    severity = round(random.uniform(0.55, 0.95), 2)

    people = random.randint(5, 60)

    children = random.randint(0, min(10, people))

    elderly = random.randint(0, min(10, people - children))

    injured = random.randint(
        max(1, int(people * 0.15)),
        max(5, int(people * 0.50))
    )

    road = "Not Accessible"

    weather = random.choice(
        ["Heavy Rain", "Thunderstorm"]
    )

    location = "Mountain"

    time_since = random.randint(5, 120)

    incident_time = random.choice(incident_times)

    infrastructure = random.choices(
        damage_levels,
        weights=[20, 40, 40]
    )[0]

    communication = random.choices(
        yes_no,
        weights=[40, 60]
    )[0]

    evacuation = random.choices(
        yes_no,
        weights=[70, 30]
    )[0]

    score, priority = calculate_priority_score(
        severity,
        people,
        injured,
        children,
        elderly,
        road,
        weather,
        time_since,
        "Landslide",
        location,
        communication,
        infrastructure,
        evacuation
    )

    return {
        "Severity": severity,
        "Disaster_Type": "Landslide",
        "People_Count": people,
        "Children_Count": children,
        "Elderly_Count": elderly,
        "Injured_Count": injured,
        "Road_Accessibility": road,
        "Weather_Condition": weather,
        "Location_Type": location,
        "Time_Since_Incident": time_since,
        "Incident_Time": incident_time,
        "Infrastructure_Damage": infrastructure,
        "Communication_Available": communication,
        "Evacuation_Required": evacuation,
        "Priority_Score": score,
        "Priority": priority
    }
```

```python
def generate_cyclone():

    severity = round(random.uniform(0.70, 1.00), 2)

    people = random.randint(50, 250)

    children = random.randint(0, min(35, people))

    elderly = random.randint(0, min(35, people - children))

    injured = random.randint(
        max(5, int(people * 0.10)),
        max(10, int(people * 0.45))
    )

    road = random.choices(
        road_accessibility,
        weights=[0, 30, 70]
    )[0]

    weather = "Cyclone"

    location = random.choice(
        ["Coastal", "Urban"]
    )

    time_since = random.randint(10, 180)

    incident_time = random.choice(incident_times)

    infrastructure = "High"

    communication = "No"

    evacuation = "Yes"

    score, priority = calculate_priority_score(
        severity,
        people,
        injured,
        children,
        elderly,
        road,
        weather,
        time_since,
        "Cyclone",
        location,
        communication,
        infrastructure,
        evacuation
    )

    return {
        "Severity": severity,
        "Disaster_Type": "Cyclone",
        "People_Count": people,
        "Children_Count": children,
        "Elderly_Count": elderly,
        "Injured_Count": injured,
        "Road_Accessibility": road,
        "Weather_Condition": weather,
        "Location_Type": location,
        "Time_Since_Incident": time_since,
        "Incident_Time": incident_time,
        "Infrastructure_Damage": infrastructure,
        "Communication_Available": communication,
        "Evacuation_Required": evacuation,
        "Priority_Score": score,
        "Priority": priority
    }
```

```python
def generate_incident():

    disaster = random.choices(
        disaster_types,
        weights=[35,20,15,15,15]
    )[0]

    if disaster == "Flood":
        return generate_flood()

    elif disaster == "Fire":
        return generate_fire()

    elif disaster == "Earthquake":
        return generate_earthquake()

    elif disaster == "Landslide":
        return generate_landslide()

    else:
        return generate_cyclone()
```

```python
data = []

for i in range(1, NUM_SAMPLES + 1):

    incident = generate_incident()

    incident["Incident_ID"] = f"INC-{i:06d}"

    data.append(incident)

print(f"Generated {len(data)} incidents.")
```

```python
df = pd.DataFrame(data)

df.head()
```
