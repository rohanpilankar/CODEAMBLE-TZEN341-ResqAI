```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
```

```python
model=YOLO("yolov8n.pt")
```

```python
image_path = r"C:\Users\HP\OneDrive\Pictures\Camera Roll\x.jpeg"
```

```python
results=model(image_path)
```

```python
annoted=results[0].plot()
plt.figure(figsize=(10,8))
plt.imshow(cv2.cvtColor(annoted,cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
people_count=0
for box in results[0].boxes:
    class_id=int(box.cls[0])
    class_name=model.names[class_id]
    confidence=float(box.conf[0])
    if class_name=="person":
        people_count+=1
        print(f"Person {people_count}: Confidence = {confidence:.2f}")
print("\nTotal Person Detected:",people_count)
```
